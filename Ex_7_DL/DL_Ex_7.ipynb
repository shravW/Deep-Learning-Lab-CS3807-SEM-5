```python
import os
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
import keras
from keras import layers, ops
from skimage.metrics import structural_similarity as ssim_fn

SEED = 42
keras.utils.set_random_seed(SEED)
rng = np.random.default_rng(SEED)

os.makedirs("figs", exist_ok=True)

def savefig(name):
    plt.tight_layout()
    plt.savefig(
        f"figs/{name}.eps",
        format="eps",
        bbox_inches="tight"
    )
    plt.show()

(x_tr_full, y_tr_full), (x_te_full, y_te_full) = keras.datasets.mnist.load_data()

x_all = (x_tr_full[:10000].astype("float32") / 255.0)[..., None]

x_train = x_all[:9000]
x_val = x_all[9000:]

x_test = (x_te_full[:2000].astype("float32") / 255.0)[..., None]
y_test = y_te_full[:2000]

print(x_train.shape, x_val.shape, x_test.shape)

flat = lambda a: a.reshape(len(a), -1)

EPOCHS = 20
BATCH = 128

def per_image_mse(x, xh):
    return np.mean((x - xh) ** 2, axis=(1, 2, 3))

def mean_ssim(x, xh):
    return float(
        np.mean([
            ssim_fn(
                a[..., 0],
                b[..., 0],
                data_range=1.0
            )
            for a, b in zip(x, xh)
        ])
    )

def evaluate(x, xh):
    return {
        "MSE": float(np.mean((x - xh) ** 2)),
        "MAE": float(np.mean(np.abs(x - xh))),
        "SSIM": mean_ssim(x, xh)
    }

def show_rows(rows, titles, n=10, name=None, cmap="gray"):
    fig, axes = plt.subplots(
        len(rows),
        n,
        figsize=(1.4 * n, 1.6 * len(rows))
    )

    for r, (imgs, title) in enumerate(zip(rows, titles)):
        for i in range(n):
            ax = axes[r, i]

            ax.imshow(
                imgs[i].squeeze(),
                cmap=cmap,
                vmin=0,
                vmax=1 if "Error" not in title else None
            )

            ax.axis("off")

            if i == 0:
                ax.set_title(
                    title,
                    fontsize=8,
                    loc="left"
                )

    if name:
        savefig(name)

results = {}

def build_fc_ae(latent=16):
    inp = keras.Input((784,))

    h = layers.Dense(
        128,
        activation="relu"
    )(inp)

    h = layers.Dense(
        32,
        activation="relu"
    )(h)

    z = layers.Dense(
        latent,
        name="latent"
    )(h)

    h = layers.Dense(
        32,
        activation="relu"
    )(z)

    h = layers.Dense(
        128,
        activation="relu"
    )(h)

    out = layers.Dense(
        784,
        activation="sigmoid"
    )(h)

    model = keras.Model(
        inp,
        out,
        name=f"fc_ae_{latent}"
    )

    model.compile(
        optimizer=keras.optimizers.Adam(1e-3),
        loss="binary_crossentropy"
    )

    return model

def train_fc(latent=16, verbose=0):
    model = build_fc_ae(latent)

    t0 = time.time()

    history = model.fit(
        flat(x_train),
        flat(x_train),
        epochs=EPOCHS,
        batch_size=BATCH,
        validation_data=(
            flat(x_val),
            flat(x_val)
        ),
        shuffle=True,
        verbose=verbose
    )

    return (
        model,
        history.history,
        time.time() - t0
    )

fc_ae, fc_hist, fc_time = train_fc(
    16,
    verbose=2
)

fc_rec = fc_ae.predict(
    flat(x_test),
    verbose=0
).reshape(
    -1,
    28,
    28,
    1
)

fc_metrics = evaluate(
    x_test,
    fc_rec
)

print(
    "FC-AE test:",
    fc_metrics
)

results["FC Autoencoder"] = {
    **fc_metrics,
    "Parameters": fc_ae.count_params(),
    "Time_s": fc_time
}

show_rows(
    [
        x_test[:10],
        fc_rec[:10],
        np.abs(
            x_test[:10] -
            fc_rec[:10]
        )
    ],
    [
        "Original",
        "Reconstruction",
        "Error |x-xhat|"
    ],
    name="plot1_fc_recon"
)

plt.figure(figsize=(6, 4))

plt.plot(
    fc_hist["loss"],
    label="Train"
)

plt.plot(
    fc_hist["val_loss"],
    label="Validation"
)

plt.xlabel("Epoch")
plt.ylabel("Reconstruction loss (BCE)")
plt.title("FC-AE loss")
plt.legend()

savefig("plot2_fc_loss")

def build_cae():
    inp = keras.Input(
        (28, 28, 1)
    )

    x = layers.Conv2D(
        32,
        3,
        activation="relu",
        padding="same"
    )(inp)

    x = layers.MaxPooling2D(
        2,
        padding="same"
    )(x)

    x = layers.Conv2D(
        64,
        3,
        activation="relu",
        padding="same"
    )(x)

    x = layers.MaxPooling2D(
        2,
        padding="same"
    )(x)

    x = layers.Conv2D(
        64,
        3,
        activation="relu",
        padding="same"
    )(x)

    x = layers.UpSampling2D(2)(x)

    x = layers.Conv2D(
        32,
        3,
        activation="relu",
        padding="same"
    )(x)

    x = layers.UpSampling2D(2)(x)

    out = layers.Conv2D(
        1,
        3,
        activation="sigmoid",
        padding="same"
    )(x)

    model = keras.Model(
        inp,
        out,
        name="cae"
    )

    model.compile(
        optimizer=keras.optimizers.Adam(1e-3),
        loss="binary_crossentropy"
    )

    return model

cae = build_cae()

t0 = time.time()

cae_hist = cae.fit(
    x_train,
    x_train,
    epochs=EPOCHS,
    batch_size=BATCH,
    validation_data=(
        x_val,
        x_val
    ),
    shuffle=True,
    verbose=2
).history

cae_time = time.time() - t0

cae_rec = cae.predict(
    x_test,
    verbose=0
)

cae_metrics = evaluate(
    x_test,
    cae_rec
)

print(
    "CAE test:",
    cae_metrics
)

results["Conv. Autoencoder"] = {
    **cae_metrics,
    "Parameters": cae.count_params(),
    "Time_s": cae_time
}

comp = pd.DataFrame({
    "Fully Connected AE": {
        **fc_metrics,
        "Parameters": fc_ae.count_params()
    },
    "Convolutional AE": {
        **cae_metrics,
        "Parameters": cae.count_params()
    }
}).T

print(comp)

show_rows(
    [
        x_test[:10],
        fc_rec[:10],
        cae_rec[:10]
    ],
    [
        "Original",
        "FC-AE",
        "CAE"
    ],
    name="plot3_fc_vs_cae"
)

def gaussian_noise(x, sigma, g):
    return np.clip(
        x + g.normal(
            0,
            sigma,
            x.shape
        ),
        0,
        1
    ).astype("float32")

def salt_pepper(x, p, g):
    m = g.random(x.shape)
    out = x.copy()

    out[m < p / 2] = 0.0

    out[
        (m >= p / 2) &
        (m < p)
    ] = 1.0

    return out.astype("float32")

def train_denoiser(noise_fn, label):
    g = np.random.default_rng(SEED)
    model = build_cae()

    history = {
        "loss": [],
        "val_loss": []
    }

    t0 = time.time()

    for ep in range(EPOCHS):
        h = model.fit(
            noise_fn(
                x_train,
                g
            ),
            x_train,
            epochs=1,
            batch_size=BATCH,
            validation_data=(
                noise_fn(
                    x_val,
                    g
                ),
                x_val
            ),
            shuffle=True,
            verbose=0
        )

        history["loss"].append(
            h.history["loss"][0]
        )

        history["val_loss"].append(
            h.history["val_loss"][0]
        )

        print(
            f"[{label}] "
            f"epoch {ep + 1}/{EPOCHS} "
            f"loss={history['loss'][-1]:.4f} "
            f"val={history['val_loss'][-1]:.4f}"
        )

    return (
        model,
        history,
        time.time() - t0
    )

dae_g, dae_g_hist, dae_g_time = train_denoiser(
    lambda x, g: gaussian_noise(
        x,
        0.2,
        g
    ),
    "DAE-Gauss(0.2)"
)

dae_sp, dae_sp_hist, dae_sp_time = train_denoiser(
    lambda x, g: salt_pepper(
        x,
        0.10,
        g
    ),
    "DAE-S&P(0.10)"
)

gen_levels = [
    0.1,
    0.2,
    0.3
]

sp_levels = [
    0.05,
    0.10,
    0.20
]

rows = []

for model_name, model in [
    (
        "DAE trained Gaussian 0.2",
        dae_g
    ),
    (
        "DAE trained S&P 0.10",
        dae_sp
    )
]:
    for noise_type, levels, fn in [
        (
            "Gaussian",
            gen_levels,
            gaussian_noise
        ),
        (
            "Salt&Pepper",
            sp_levels,
            salt_pepper
        )
    ]:
        for level in levels:
            g = np.random.default_rng(1000)

            xn = fn(
                x_test,
                level,
                g
            )

            xd = model.predict(
                xn,
                verbose=0
            )

            den = evaluate(
                x_test,
                xd
            )

            noisy = evaluate(
                x_test,
                xn
            )

            rows.append({
                "Model": model_name,
                "Noise": noise_type,
                "Level": level,
                "MSE": den["MSE"],
                "MAE": den["MAE"],
                "SSIM": den["SSIM"],
                "Noisy_MSE": noisy["MSE"],
                "Noisy_MAE": noisy["MAE"],
                "Noisy_SSIM": noisy["SSIM"]
            })

noise_df = pd.DataFrame(rows)

print(
    noise_df.round(4).to_string(
        index=False
    )
)

noise_df.to_csv(
    "noise_results.csv",
    index=False
)

g = np.random.default_rng(7)

xn_g = gaussian_noise(
    x_test[:10],
    0.2,
    g
)

show_rows(
    [
        x_test[:10],
        xn_g,
        dae_g.predict(
            xn_g,
            verbose=0
        )
    ],
    [
        "Clean",
        "Noisy (Gauss 0.2)",
        "Denoised"
    ],
    name="plot4_denoise_gaussian"
)

xn_s = salt_pepper(
    x_test[:10],
    0.10,
    g
)

show_rows(
    [
        x_test[:10],
        xn_s,
        dae_sp.predict(
            xn_s,
            verbose=0
        )
    ],
    [
        "Clean",
        "Noisy (S&P 0.10)",
        "Denoised"
    ],
    name="plot4_denoise_saltpepper"
)

sub = noise_df[
    (noise_df.Model == "DAE trained Gaussian 0.2") &
    (noise_df.Noise == "Gaussian")
]

fig, axes = plt.subplots(
    1,
    3,
    figsize=(13, 3.6)
)

for ax, metric in zip(
    axes,
    [
        "MSE",
        "MAE",
        "SSIM"
    ]
):
    ax.plot(
        sub.Level,
        sub[metric],
        "o-",
        label="Denoised"
    )

    ax.plot(
        sub.Level,
        sub["Noisy_" + metric],
        "s--",
        label="Noisy input"
    )

    ax.set_xlabel(
        "Gaussian sigma"
    )

    ax.set_ylabel(metric)
    ax.set_title(metric)
    ax.legend()

savefig(
    "plot5_noise_vs_quality"
)

dae_eval = sub[
    sub.Level == 0.2
].iloc[0]

results["Denoising CAE"] = {
    "MSE": dae_eval.MSE,
    "MAE": dae_eval.MAE,
    "SSIM": dae_eval.SSIM,
    "Parameters": dae_g.count_params(),
    "Time_s": dae_g_time
}

LATENT = 2

class Sampling(layers.Layer):
    def call(self, inputs):
        z_mean, z_log_var = inputs

        batch = ops.shape(
            z_mean
        )[0]

        dim = ops.shape(
            z_mean
        )[1]

        eps = keras.random.normal(
            shape=(batch, dim)
        )

        return (
            z_mean +
            ops.exp(
                0.5 * z_log_var
            ) * eps
        )

def build_vae_parts(latent=LATENT):
    enc_in = keras.Input(
        (28, 28, 1)
    )

    x = layers.Conv2D(
        32,
        3,
        strides=2,
        padding="same",
        activation="relu"
    )(enc_in)

    x = layers.Conv2D(
        64,
        3,
        strides=2,
        padding="same",
        activation="relu"
    )(x)

    x = layers.Flatten()(x)

    x = layers.Dense(
        16,
        activation="relu"
    )(x)

    z_mean = layers.Dense(
        latent,
        name="z_mean"
    )(x)

    z_log_var = layers.Dense(
        latent,
        name="z_log_var"
    )(x)

    z = Sampling()([
        z_mean,
        z_log_var
    ])

    encoder = keras.Model(
        enc_in,
        [
            z_mean,
            z_log_var,
            z
        ],
        name="encoder"
    )

    dec_in = keras.Input(
        (latent,)
    )

    x = layers.Dense(
        7 * 7 * 64,
        activation="relu"
    )(dec_in)

    x = layers.Reshape(
        (7, 7, 64)
    )(x)

    x = layers.Conv2DTranspose(
        64,
        3,
        strides=2,
        padding="same",
        activation="relu"
    )(x)

    x = layers.Conv2DTranspose(
        32,
        3,
        strides=2,
        padding="same",
        activation="relu"
    )(x)

    dec_out = layers.Conv2DTranspose(
        1,
        3,
        padding="same",
        activation="sigmoid"
    )(x)

    decoder = keras.Model(
        dec_in,
        dec_out,
        name="decoder"
    )

    return encoder, decoder

class VAE(keras.Model):
    def __init__(
        self,
        encoder,
        decoder,
        beta=1.0,
        **kwargs
    ):
        super().__init__(
            **kwargs
        )

        self.encoder = encoder
        self.decoder = decoder
        self.beta = beta

        self.total_tracker = keras.metrics.Mean(
            name="loss"
        )

        self.rec_tracker = keras.metrics.Mean(
            name="rec_loss"
        )

        self.kl_tracker = keras.metrics.Mean(
            name="kl_loss"
        )

    @property
    def metrics(self):
        return [
            self.total_tracker,
            self.rec_tracker,
            self.kl_tracker
        ]

    def call(self, x):
        return self.decoder(
            self.encoder(x)[2]
        )

    def compute_losses(self, x):
        z_mean, z_log_var, z = self.encoder(x)

        xh = self.decoder(z)

        rec = ops.mean(
            ops.sum(
                keras.losses.binary_crossentropy(
                    x,
                    xh
                ),
                axis=(1, 2)
            )
        )

        kl = ops.mean(
            ops.sum(
                -0.5 * (
                    1 +
                    z_log_var -
                    ops.square(z_mean) -
                    ops.exp(z_log_var)
                ),
                axis=1
            )
        )

        return rec, kl

    def _update(self, rec, kl):
        total = (
            rec +
            self.beta * kl
        )

        self.total_tracker.update_state(
            total
        )

        self.rec_tracker.update_state(
            rec
        )

        self.kl_tracker.update_state(
            kl
        )

        return {
            metric.name: metric.result()
            for metric in self.metrics
        }

    def train_step(self, data):
        x = (
            data[0]
            if isinstance(
                data,
                (tuple, list)
            )
            else data
        )

        with tf.GradientTape() as tape:
            rec, kl = self.compute_losses(x)
            total = (
                rec +
                self.beta * kl
            )

        grads = tape.gradient(
            total,
            self.trainable_weights
        )

        self.optimizer.apply_gradients(
            zip(
                grads,
                self.trainable_weights
            )
        )

        return self._update(
            rec,
            kl
        )

    def test_step(self, data):
        x = (
            data[0]
            if isinstance(
                data,
                (tuple, list)
            )
            else data
        )

        rec, kl = self.compute_losses(x)

        return self._update(
            rec,
            kl
        )

encoder, decoder = build_vae_parts()

vae = VAE(
    encoder,
    decoder,
    beta=1.0
)

vae.compile(
    optimizer=keras.optimizers.Adam(
        1e-3
    )
)

VAE_EPOCHS = 30

t0 = time.time()

vae_hist = vae.fit(
    x_train,
    epochs=VAE_EPOCHS,
    batch_size=BATCH,
    validation_data=(x_val,),
    shuffle=True,
    verbose=2
).history

vae_time = time.time() - t0

z_mean_te, z_logvar_te, _ = encoder.predict(
    x_test,
    verbose=0
)

vae_rec_img = decoder.predict(
    z_mean_te,
    verbose=0
)

rec_l, kl_l = [
    float(v)
    for v in vae.compute_losses(
        x_test
    )
]

vae_metrics = evaluate(
    x_test,
    vae_rec_img
)

vae_table = pd.Series({
    "Reconstruction Loss": rec_l,
    "KL Loss": kl_l,
    "Total Loss": rec_l + kl_l,
    "Test MSE": vae_metrics["MSE"],
    "Test MAE": vae_metrics["MAE"],
    "Mean SSIM": vae_metrics["SSIM"]
})

print(vae_table)

results["VAE"] = {
    **vae_metrics,
    "Parameters": (
        encoder.count_params() +
        decoder.count_params()
    ),
    "Time_s": vae_time
}

fig, axes = plt.subplots(
    1,
    3,
    figsize=(13, 3.6)
)

for ax, key, title in zip(
    axes,
    [
        "rec_loss",
        "kl_loss",
        "loss"
    ],
    [
        "Reconstruction loss",
        "KL loss",
        "Total loss"
    ]
):
    ax.plot(
        vae_hist[key],
        label="Train"
    )

    ax.plot(
        vae_hist["val_" + key],
        label="Validation"
    )

    ax.set_xlabel("Epoch")
    ax.set_ylabel(title)
    ax.set_title(title)
    ax.legend()

savefig(
    "plot9_vae_loss"
)

plt.figure(
    figsize=(6.5, 5.5)
)

sc = plt.scatter(
    z_mean_te[:, 0],
    z_mean_te[:, 1],
    c=y_test,
    cmap="tab10",
    s=8
)

plt.colorbar(
    sc,
    ticks=range(10),
    label="Digit"
)

plt.xlabel("z1")
plt.ylabel("z2")

plt.title(
    "VAE latent space (test set, posterior means)"
)

savefig(
    "plot6_vae_latent"
)

z_rand = np.random.default_rng(
    SEED
).normal(
    size=(25, LATENT)
).astype("float32")

gen = decoder.predict(
    z_rand,
    verbose=0
)

fig, axes = plt.subplots(
    5,
    5,
    figsize=(6, 6)
)

for ax, im in zip(
    axes.ravel(),
    gen
):
    ax.imshow(
        im.squeeze(),
        cmap="gray"
    )

    ax.axis("off")

plt.suptitle(
    "VAE generated samples"
)

savefig(
    "plot7_vae_generated"
)

DIGIT_A = 3
DIGIT_B = 8

zA = z_mean_te[
    np.where(
        y_test == DIGIT_A
    )[0][0]
]

zB = z_mean_te[
    np.where(
        y_test == DIGIT_B
    )[0][0]
]

alphas = np.linspace(
    0,
    1,
    11
)

z_int = np.array(
    [
        (1 - a) * zA + a * zB
        for a in alphas
    ],
    dtype="float32"
)

interp = decoder.predict(
    z_int,
    verbose=0
)

fig, axes = plt.subplots(
    1,
    11,
    figsize=(15, 1.9)
)

for ax, im, alpha in zip(
    axes,
    interp,
    alphas
):
    ax.imshow(
        im.squeeze(),
        cmap="gray"
    )

    ax.axis("off")

    ax.set_title(
        f"{alpha:.1f}",
        fontsize=8
    )

plt.suptitle(
    f"Interpolation: digit {DIGIT_A} -> digit {DIGIT_B}",
    y=1.05
)

savefig(
    "plot8_interpolation"
)

n = 15
lim = 3.0

grid = np.linspace(
    -lim,
    lim,
    n
)

canvas = np.zeros(
    (28 * n, 28 * n)
)

for i, yv in enumerate(
    grid[::-1]
):
    for j, xv in enumerate(grid):
        decoded = decoder.predict(
            np.array(
                [[xv, yv]],
                dtype="float32"
            ),
            verbose=0
        )

        canvas[
            i * 28:(i + 1) * 28,
            j * 28:(j + 1) * 28
        ] = decoded[
            0,
            ...,
            0
        ]

plt.figure(
    figsize=(7, 7)
)

plt.imshow(
    canvas,
    cmap="gray"
)

plt.axis("off")

plt.title(
    "Decoded latent manifold"
)

savefig(
    "bonus_vae_manifold"
)

def error_analysis(x, xh, tag):
    e = per_image_mse(
        x,
        xh
    )

    print(
        f"[{tag}] "
        f"error mean={e.mean():.5f} "
        f"median={np.median(e):.5f} "
        f"std={e.std():.5f} "
        f"p95={np.percentile(e, 95):.5f} "
        f"max={e.max():.5f}"
    )

    plt.figure(
        figsize=(6, 4)
    )

    plt.hist(
        e,
        bins=50
    )

    plt.axvline(
        e.mean(),
        color="r",
        ls="--",
        label="mean"
    )

    plt.xlabel(
        "Per-image MSE"
    )

    plt.ylabel(
        "Count"
    )

    plt.title(
        f"Reconstruction error - {tag}"
    )

    plt.legend()

    savefig(
        f"plot10_error_hist_{tag}"
    )

    top = np.argsort(e)[-5:][::-1]

    show_rows(
        [
            x[top],
            xh[top]
        ],
        [
            "Original (top-5 error)",
            "Reconstruction"
        ],
        n=5,
        name=f"high_error_{tag}"
    )

    print(
        "Top-5 indices:",
        top,
        "digits:",
        y_test[top],
        "errors:",
        np.round(
            e[top],
            4
        )
    )

error_analysis(
    x_test,
    fc_rec,
    "FC_AE"
)

error_analysis(
    x_test,
    cae_rec,
    "CAE"
)

lat_rows = []

for dz in [
    2,
    8,
    16,
    32
]:
    model, _, _ = train_fc(dz)

    rec = model.predict(
        flat(x_test),
        verbose=0
    ).reshape(
        -1,
        28,
        28,
        1
    )

    metrics = evaluate(
        x_test,
        rec
    )

    lat_rows.append({
        "LatentDim": dz,
        "MSE": metrics["MSE"],
        "SSIM": metrics["SSIM"],
        "Parameters": model.count_params()
    })

lat_df = pd.DataFrame(
    lat_rows
)

print(lat_df)

plt.figure(
    figsize=(5.5, 4)
)

plt.plot(
    lat_df.LatentDim,
    lat_df.MSE,
    "o-"
)

plt.xscale(
    "log",
    base=2
)

plt.xticks(
    lat_df.LatentDim,
    lat_df.LatentDim
)

plt.xlabel(
    "Latent dimension"
)

plt.ylabel(
    "Test MSE"
)

plt.title(
    "Latent dimension vs MSE"
)

savefig(
    "plot10_latent_dim_vs_mse"
)

cons = pd.DataFrame(
    results
).T[
    [
        "MSE",
        "MAE",
        "SSIM",
        "Parameters",
        "Time_s"
    ]
]

print(
    cons.round(5)
)

cons.to_csv(
    "consolidated_results.csv"
)

comp.to_csv(
    "fc_vs_cae.csv"
)

vae_table.to_csv(
    "vae_metrics.csv"
)

lat_df.to_csv(
    "latent_dim_study.csv"
)

print("\nSaved EPS plots:")

for file in sorted(
    os.listdir("figs")
):
    if file.endswith(".eps"):
        print(
            f"figs/{file}"
        )
```
